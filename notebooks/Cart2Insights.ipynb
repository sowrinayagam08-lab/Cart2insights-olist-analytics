```python
import re  # The function uses it to cut city names at /, , and -
import unicodedata
```

```python
# step 7 -> Write a cleaning function
def clean_city(s):
    # remove accents, lowercase, trim spaces
    s = unicodedata.normalize("NFKD", s).encode("ascii", "ignore").decode().lower().strip()

    # values that are not cities at all (zip codes, emails)
    if s.isdigit() or "@" in s:
        return None

    # keep only the part before  /  \  ,  or  " - "
    s = re.split(r"\s*[/\\,]\s*|\s+-\s+", s)[0]

    # remove text in brackets, e.g. "(porto seguro)"
    s = re.sub(r"\s*\(.*\)", "", s)

    # remove a state attached with a hyphen, e.g. "andira-pr"
    s = re.sub(r"-[a-z]{2}$", "", s)

    # fix known abbreviations
    return {"sbc": "sao bernardo do campo", "sp": "sao paulo"}.get(s.strip(), s.strip())
```

```python
"""
Each line handles one kind of mess:

1.normalize(...) removes accents (são becomes sao), then .lower().strip() handles case and spaces.
2.isdigit() and "@" catch the zip code and email. They aren't cities, so the function returns None (missing).
3.re.split(...) cuts the text at /, \, , or a spaced hyphen and keeps the first piece. novo hamburgo, rio grande do sul, brasil becomes novo hamburgo.
4.The last regex removes a hyphen followed by exactly two letters at the end. I only cut " - " with spaces in Step 4's split, 
so real hyphenated city names like embu-guacu stay safe, except a name ending in a two-letter piece, so check Step 6.
5.The small dictionary expands abbreviations.

"""
```

```python
# step 8 -> Apply it
df7["seller_city"] = df7["seller_city"].apply(clean_city)
print("Missing cities:", df7["seller_city"].isnull().sum())
df7["seller_city"] = df7["seller_city"].fillna("unknown")
```

```python
# Check the result
print("Cities:", df7["seller_city"].nunique())
odd = df7[df7["seller_city"].str.contains(r"[0-9@/\\,;()\-\.]")]
print(odd["seller_city"].unique())

"""
The city count drops from 611 to about 587 (586 real cities plus unknown), and the odd-character list should be empty. 
If a real hyphenated city appears in the list, tell me and we'll add it as an exception.
"""
```
